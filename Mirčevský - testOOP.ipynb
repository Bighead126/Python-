# 2
Vytvořte třídu `Knihovna` s těmito atributy (nastavte v `__init__`): **(1b)**
* `nazev`
* `knihy` — seznam knih (výchozí prázdný seznam)
* `kapacita` — maximální počet knih

Implementujte tyto metody: **(1b + 2b)**

* `__str__` — vrátí přehled o knihovně (název, obsazenost, výpis knih)
* `__len__` — vrátí aktuální počet knih
* `__contains__(self, kniha)` — umožní použití operátoru `in` (porovnání podle názvu)
* `pridej_knihu(self, kniha)` a `odeber_knihu(self, kniha)` — přidání/odebrání knihy s kontrolou kapacity a duplicit + zpětná vazba pomocí `print()` **(+2b)**

````python
#TODO class Knihovna
class Knihovna:
    def __init__(self, nazev, kapacita, knihy=None):
        self.nazev = nazev
        self.kapacita = kapacita
        self.knihy = knihy if knihy is not None else []

    def __str__(self):
        vystup = f"Knihovna: {self.nazev}\n"
        vystup += f"Obsazenost: {len(self.knihy)}/{self.kapacita}\n"
        vystup += "Seznam knih:\n"
        
        if not self.knihy:
            vystup += "  (zadne knihy)\n"
        else:
            for kniha in self.knihy:
                vystup += f"  - {kniha}\n"
        
        return vystup

    def __len__(self):
        return len(self.knihy)

    def __contains__(self, kniha):
        return kniha in self.knihy

    def pridej_knihu(self, kniha):
        if len(self.knihy) >= self.kapacita:
            print(f"Nelze pridat '{kniha}': kapacita je naplnena.")
        elif kniha in self.knihy:
            print(f"Kniha '{kniha}' už v knihovne existuje.")
        else:
            self.knihy.append(kniha)
            print(f"Kniha '{kniha}' byla pridana.")

    def odeber_knihu(self, kniha):
        if kniha in self.knihy:
            self.knihy.remove(kniha)
            print(f"Kniha '{kniha}' byla odebrana.")
        else:
            print(f"Kniha '{kniha}' v knihovne neni.")



k = Knihovna("Mestská knihovna", 3)

k.pridej_knihu("1984")
k.pridej_knihu("Maly princ")
k.pridej_knihu("Harry Potter")
k.pridej_knihu("Pán prstenu")  

print()
print(k)

print("Pocet knih:", len(k))

print("Je '1984' v knihovne?", "1984" in k)

k.odeber_knihu("1984")
print()
print(k)

        
````
